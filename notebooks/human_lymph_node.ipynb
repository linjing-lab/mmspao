```python
from mmspao.utils import *
from mmspao import mmspao
import scanpy as sc

set_random_seed(42)
device = 'cuda'
```

```python
#Load data and perform necessary preprocessing
rna = sc.read('/home/data1/linjing0806/multiomics/SpatialGlue/data/Dataset11_Human_Lymph_Node_A1/adata_RNA.h5ad')
protein = sc.read('/home/data1/linjing0806/multiomics/SpatialGlue/data/Dataset11_Human_Lymph_Node_A1/adata_ADT.h5ad')
sc.pp.highly_variable_genes(rna, flavor="seurat_v3", n_top_genes=3000)
sc.pp.normalize_total(rna, target_sum=1e4)
rna =  rna[:, rna.var['highly_variable']]
adata = rna.copy()
rna = getX(rna,modality='rna')
```

```python
protein = getX(protein,modality='protein')
```

```python
model = mmspao([rna,protein],sel_views='all',combination='all', sparse=True, device=device)
model.train()
```

```python
model.cluster(adata, n_domains=6, end=1.0)
```

```python
sc.pl.spatial(adata, color="leiden", spot_size=1.5, show=True)
```

```python
sc.tl.umap(adata)
sc.pl.umap(adata, color="leiden",)
```
